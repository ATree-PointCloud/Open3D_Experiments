```python
import numpy as np
import open3d as o3d
print(o3d.__version__)
```

### 1. 点云可视化(示例文件)

```python
# load 一帧内置点云数据示例
ply_point_cloud  = o3d.data.PLYPointCloud()

# 通过给定路径读取一帧点云
pcd = o3d.io.read_point_cloud(ply_point_cloud.path)
print(pcd)
print(np.asarray(pcd.points).shape)

# 可视化
o3d.visualization.draw_geometries([pcd],
                                  zoom=0.3412,
                                  front=[0.4257, -0.2125, -0.8795],
                                  lookat=[2.6172, 2.0475, 1.532],
                                  up=[-0.0694, -0.9768, 0.2024])
```

### 2. 体素下采样 (Voxel Downsampling)
体素下采样利用规则的体素网格，将输入的点云转换为经过均匀下采样的点云。该技术常被用作各种点云处理任务的预处理步骤。该算法分为两个步骤来执行:
- 将点分配到各个体素中。
- 每个体素内所有点进行平均值计算，从而产生一个点。

```python
downpcd = pcd.voxel_down_sample(voxel_size=0.02)
o3d.visualization.draw_geometries([downpcd],
                                  zoom=0.3412,
                                  front=[0.4257, -0.2125, -0.8795],
                                  lookat=[2.6172, 2.0475, 1.532],
                                  up=[-0.0694, -0.9768, 0.2024])
```

### 3. 点云去噪

#### 3.1 统计滤波
 - nb_neighbors：用于计算均值的邻域点数
 - std_ratio：标准差倍数阈值，超过则视为离群点

```python
filtered_pcd, ind = pcd.remove_statistical_outlier(nb_neighbors=20, std_ratio=2.0)
outliers = pcd.select_by_index(ind, invert = True)

# filtered_pcd.paint_uniform_color([0, 1, 0])
outliers.paint_uniform_color([1, 0, 0])
o3d.visualization.draw_geometries([filtered_pcd, outliers],
                                  zoom=0.3412,
                                  front=[0.4257, -0.2125, -0.8795],
                                  lookat=[2.6172, 2.0475, 1.532],
                                  up=[-0.0694, -0.9768, 0.2024])
```
